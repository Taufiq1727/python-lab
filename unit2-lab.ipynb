### program to print the number of characters in a string

```python
text = "Taufiq"
frequency = {}

for ch in text:
    if ch in frequency:
        frequency[ch] = +1
    else:
        frequency[ch] = 1
print(frequency)
```

```python
# Write a program to calculate factorial of a number using recursion 

def fact(num):
    if num == 0 or num == 1:
        return 1
    else:
        return num*fact(num-1)

num = int(input("Enter a number:"))
res = fact(num)
print(f"The factorial of {num} is {res}")
```

```python
# find if the word is palindrome or not

def palindrome(word):
    if word[::-1] == word:
        print("It is a palindrome")
    else:
        print("It is not a palindrome")


word = input("Enter the word")
palindrome(word)

```

```python
# Python program to check whether a string is palindrome or not using recursion

def palindrome(text):
    
    # Base condition
    if len(text) <= 1:
        return True
    
    # Check first and last characters
    if text[0] == text[-1]:
        return palindrome(text[1:-1])
    
    return False


# Taking input from user
text = input("Enter a string: ")

# Checking palindrome
if palindrome(text):
    print("It is a Palindrome")
else:
    print("It is Not a Palindrome")
```
