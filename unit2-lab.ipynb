LISTS

```python
lst=[1,2,3,4,5,6]
print(lst)
lst.append(6)
print(lst)
lst.remove(4)
print(lst)
print(lst[0])
```

Sorting the elements in the list

```python
list = [5,6,8,2,3]
list.sort()
print(list)
```

```python
list = [0,6,7,8,5]
list.reverse()
print(list)
```

## making new list and doing extra operations

```python
list = ["apple", "banana", "mango"]
print(list)
```

```python
#add element 
list.append("orange")
print(list)
```

```python
#remove element
if"banana" in list:
    list.remove("banana")

print(list)
```

```python
#access by index
print("first fruit:",list[0])
```

```python
#list comprehension
upper_fruits = [list.upper() for list in list]
print("Upper case fruit:", upper_fruits)
```

### 2 Tuples(Immutable,ordered,allows,duplicates)

```python
print("\n----------tuple example------------\n")
coordinates = (10.5,20.3,10.3)
print("Tuple", coordinates)
print("first coordinate:",coordinates[0])
```

### Tuple unpacking


```python
x,y,z = coordinates
print(f"Unpacked: x={x}, y={y},z={z}")
```

## SETS


```python
print("\n----set sample----\n")
numbers = {1,2,3,4,1}
print("set:",numbers)
numbers.add(4)
numbers.add(7)
numbers.discard(0)
print(numbers)
```
