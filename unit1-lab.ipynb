## 1a & 1b: Input Name and Age

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
print("Name:", name)
print("Age:", age)
```

## 1c: Arithmetic Operations

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

## 1d: Relational and Logical Operators

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("a > b:", a > b)
print("a < b:", a < b)
print("a == b:", a == b)
print("a != b:", a != b)

print("Logical AND:", a > 0 and b > 0)
print("Logical OR:", a > 0 or b > 0)
```

## 1e: Bitwise Operators

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```

## 1f: Assignment Operators

```python
a = 10
print("Initial:", a)

a += 5
print("a += 5:", a)

a -= 3
print("a -= 3:", a)

a *= 2
print("a *= 2:", a)

a /= 4
print("a /= 4:", a)
```

## 1g: Membership in List

```python
lst = [1, 2, 3, 4, 5]
num = int(input("Enter number: "))

print(num in lst)
```

## 1h: Membership in String

```python
text = "hello world"
ch = input("Enter character: ")

print(ch in text)
```

## 1i: Identity Operators with Integers

```python
a = 10
b = 10

print("a is b:", a is b)
print("a is not b:", a is not b)
```

## 1j: Identity Operators with Lists

```python
a = [1, 2, 3]
b = [1, 2, 3]

print("a is b:", a is b)
print("a == b:", a == b)
```

## 2a: Even or Odd

```python
num = int(input("Enter number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

## 2b: Positive, Negative or Zero

```python
num = int(input("Enter number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

## 2c: Largest of Three Numbers

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a >= b and a >= c:
    print("Largest:", a)
elif b >= a and b >= c:
    print("Largest:", b)
else:
    print("Largest:", c)
```

## 2d: While Loop (1 to 5)

```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

## 2e: For Loop (1 to 5)

```python
for i in range(1, 6):
    print(i)
```

## 2f: Sum of N Natural Numbers

```python
n = int(input("Enter N: "))
sum = 0

for i in range(1, n+1):
    sum += i

print("Sum:", sum)
```

## 2g: Multiplication Table

```python
n = int(input("Enter number: "))

for i in range(1, 11):
    print(n, "x", i, "=", n*i)
```

## 2h: Count Digits

```python
num = int(input("Enter number: "))
count = 0

while num != 0:
    num //= 10
    count += 1

print("Digits:", count)
```

## 2i: Factorial

```python
n = int(input("Enter number: "))
fact = 1

for i in range(1, n+1):
    fact *= i

print("Factorial:", fact)
```
